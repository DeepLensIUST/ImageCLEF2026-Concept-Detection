### Fix 'ID order mismatch' error

The submission system expects the entries in the CSV file to be sorted by the 'ID' column.

```python
import pandas as pd

input_csv_path = 'submission.csv' 
output_csv_path = 'submission_sorted.csv'

# --- Load the CSV file ---
try:
    df = pd.read_csv(input_csv_path)
    print(f"Successfully loaded '{input_csv_path}'. First 5 rows:")
    display(df.head())

    # --- Sort the DataFrame by the ID column ---
    if 'ID' in df.columns:
        # Extract the numeric part of the ID for proper sorting
        df['sort_key'] = df['ID'].str.extract(r'(\d+)$').astype(int)
        df_sorted = df.sort_values(by='sort_key').drop(columns=['sort_key'])
        print("\nDataFrame sorted by 'ID' column. First 5 rows of sorted data:")
        display(df_sorted.head())

        # --- Save the sorted DataFrame to a new CSV file ---
        df_sorted.to_csv(output_csv_path, index=False)
        print(f"\nSorted submission saved to '{output_csv_path}'.")
        print("You can now download this file and use it for your competition submission.")
    else:
        print("Error: 'ID' column not found in the CSV. Please check your CSV file and update the column name if necessary.")

except FileNotFoundError:
    print(f"Error: The file '{input_csv_path}' was not found. Please ensure the file is uploaded to the correct path and update 'input_csv_path' if needed.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
